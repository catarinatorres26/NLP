```python
path = "project/data/train.jsonl"

bad = 0
with open(path, "r", encoding="utf-8") as f:
    for i, line in enumerate(f, start=1):
        try:
            obj = json.loads(line)
            assert "instruction" in obj
            assert "response" in obj
            assert "short_answer" in obj["response"]
            assert "confidence_level" in obj["response"]
        except Exception as e:
            bad += 1
            print(f"Bad line {i}: {e}")

print("Total lines:", sum(1 for _ in open(path, 'r', encoding='utf-8')))
print("Bad lines:", bad)
```

```python
import json, re, random, hashlib
from copy import deepcopy

# Junta aqui todas as listas que já tens no notebook.
# Exemplo: seed = examples + more + more2 + more3 ...
seed = []
for name in ["examples", "more"]:
    if name in globals() and isinstance(globals()[name], list):
        seed += [x for x in globals()[name] if isinstance(x, dict)]


assert len(seed) > 0, "Seed vazio: confirma que tens 'examples'/'more' definidos antes desta célula."

def canonicalize_text(s: str) -> str:
    s = s.strip()
    s = re.sub(r"\s+", " ", s)
    return s

def validate_ex(ex):
    assert "instruction" in ex
    assert "response" in ex and isinstance(ex["response"], dict)
    assert "short_answer" in ex["response"]
    assert "confidence_level" in ex["response"]
    ex["instruction"] = canonicalize_text(ex["instruction"])
    ex["response"]["short_answer"] = canonicalize_text(ex["response"]["short_answer"])
    if "clinical_notes" in ex["response"] and ex["response"]["clinical_notes"] is not None:
        ex["response"]["clinical_notes"] = canonicalize_text(ex["response"]["clinical_notes"])
    return ex

seed = [validate_ex(deepcopy(x)) for x in seed]
print("Seed size:", len(seed))
print("Exemplo:", seed[0]["instruction"])
```

```python
PARAPHRASE_PATTERNS = [
    "In clinical terms, {q}",
    "Briefly: {q}",
    "As a clinician, how would you answer: {q}",
    "What’s the evidence-based answer to: {q}",
    "Explain in one paragraph: {q}",
    "For patient counseling, {q}",
    "Considering safety and regulatory boundaries, {q}",
    "From a mechanistic perspective, {q}",
]

def paraphrase_instruction(q: str, k: int = 4):
    q = canonicalize_text(q)
    random.shuffle(PARAPHRASE_PATTERNS)
    out = []
    for pat in PARAPHRASE_PATTERNS[:k]:
        out.append(canonicalize_text(pat.format(q=q)))
    return out
```

```python
VIGNETTES = [
    "A 62-year-old with type 2 diabetes and ASCVD asks: {q}",
    "A patient on GLP-1 RA reports nausea in week 2. They ask: {q}",
    "A patient with suspected gastroparesis asks: {q}",
    "An overweight patient considering off-label use for longevity asks: {q}",
    "A clinician is documenting informed consent and asks: {q}",
    "A frail older adult is losing weight rapidly. The team asks: {q}",
    "A patient with family history of MTC asks: {q}",
]

def vignette_instruction(q: str, k: int = 3):
    q = canonicalize_text(q)
    v = VIGNETTES[:]
    random.shuffle(v)
    return [canonicalize_text(v[i].format(q=q)) for i in range(min(k, len(v)))]
```

```python
SAFETY_FRAMES = [
    "If evidence is limited, say so. {q}",
    "Answer conservatively and include safety notes if relevant: {q}",
    "Do not claim approval unless explicitly approved. {q}",
    "Differentiate clinical evidence vs mechanistic hypotheses: {q}",
]

def safety_instruction(q: str, k: int = 2):
    q = canonicalize_text(q)
    frames = SAFETY_FRAMES[:]
    random.shuffle(frames)
    return [canonicalize_text(frames[i].format(q=q)) for i in range(min(k, len(frames)))]
```

```python
def ex_key(ex):
    # chave estável por instruction + short_answer (evita duplicados óbvios)
    m = hashlib.md5()
    m.update((ex["instruction"] + "||" + ex["response"]["short_answer"]).encode("utf-8"))
    return m.hexdigest()

def expand(seed, p=4, v=3, s=2):
    out = []
    for base in seed:
        out.append(deepcopy(base))

        q = base["instruction"]
        variants = []
        variants += paraphrase_instruction(q, k=p)
        variants += vignette_instruction(q, k=v)
        variants += safety_instruction(q, k=s)

        for instr in variants:
            ex = deepcopy(base)
            ex["instruction"] = instr
            out.append(validate_ex(ex))
    # dedupe
    seen = set()
    deduped = []
    for ex in out:
        k = ex_key(ex)
        if k not in seen:
            seen.add(k)
            deduped.append(ex)
    return deduped

expanded = expand(seed, p=4, v=3, s=2)
print("Expanded (deduped):", len(expanded))

TARGET = 310
random.seed(42)
random.shuffle(expanded)

if len(expanded) < TARGET:
    raise ValueError(f"Não chega a {TARGET}. Tens {len(expanded)}. Aumenta p/v/s ou adiciona mais seed canónico.")
final = expanded[:TARGET]
print("Final:", len(final))
```

```python
import os, json

os.makedirs("project/data", exist_ok=True)
path = "project/data/train.jsonl"

with open(path, "w", encoding="utf-8") as f:
    for ex in final:
        f.write(json.dumps(ex, ensure_ascii=False) + "\n")

# validação rápida
bad = 0
with open(path, "r", encoding="utf-8") as f:
    for i, line in enumerate(f, start=1):
        try:
            obj = json.loads(line)
            assert "instruction" in obj
            assert "response" in obj
            assert "short_answer" in obj["response"]
            assert obj["response"]["confidence_level"] in ["high","medium","low"]
        except Exception as e:
            bad += 1
            print("Bad line", i, e)

print("Wrote:", path)
print("Total lines:", sum(1 for _ in open(path, "r", encoding="utf-8")))
print("Bad lines:", bad)
```

```python
import json, random, os

random.seed(42)

TRAIN_PATH = "project/data/train.jsonl"
DEV_PATH   = "project/data/dev.jsonl"
TARGET_DEV = 30

# 1) Ler train atual (não mexe nele)
with open(TRAIN_PATH, "r", encoding="utf-8") as f:
    train = [json.loads(line) for line in f if line.strip()]

train_instructions = set(ex["instruction"].strip() for ex in train)
print("Train examples:", len(train))
print("Unique train instructions:", len(train_instructions))

# 2) Construir um pool de candidatos para dev a partir do seed (sem depender do teu test final)
#    - usa as tuas bases (seed) e cria variações leves
assert "seed" in globals(), "Não existe 'seed' no notebook. Corre primeiro a célula onde constróis o seed."

# Se ainda não tens expand() definido, não precisamos dele: vamos só usar as próprias bases (seed).
# Mas se tiveres expand(), podemos aumentar o pool para garantir 30 fora do treino.
pool = []

if "expand" in globals():
    # pequeno augment para ter pool suficiente
    pool = expand(seed, p=2, v=1, s=1)
else:
    pool = seed[:]  # só bases

# 3) Filtrar candidatos que NÃO aparecem no train (por instruction)
candidates = [ex for ex in pool if ex["instruction"].strip() not in train_instructions]

# 4) Garantir 30 instructions únicas no dev
random.shuffle(candidates)

dev = []
seen_dev_instr = set()
for ex in candidates:
    instr = ex["instruction"].strip()
    if instr in seen_dev_instr:
        continue
    seen_dev_instr.add(instr)
    dev.append(ex)
    if len(dev) == TARGET_DEV:
        break

if len(dev) < TARGET_DEV:
    raise ValueError(f"Não consegui obter {TARGET_DEV} exemplos únicos fora do train. Consegui {len(dev)}. "
                     f"Isto significa que o teu train já cobre quase tudo o que existe no pool.")

# 5) Escrever dev.jsonl (só isto muda)
os.makedirs("project/data", exist_ok=True)
with open(DEV_PATH, "w", encoding="utf-8") as f:
    for ex in dev:
        f.write(json.dumps(ex, ensure_ascii=False) + "\n")

print("Wrote:", DEV_PATH, "with", len(dev), "examples")

```
